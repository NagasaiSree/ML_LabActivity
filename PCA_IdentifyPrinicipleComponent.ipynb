```python
import pandas as pd
import matplotlib.pyplot as plt

loadings = pd.DataFrame({
    'PC1': [0.190, 0.544, 0.782, 0.365, 0.585, 0.394, 0.985, 0.520, 0.142],
    'PC2': [0.017, 0.020, -0.605, 0.294, 0.085, -0.273, 0.126, 0.402, 0.150],
    'PC3': [0.207, 0.204, 0.144, 0.585, 0.234, 0.027, -0.111, 0.519, 0.239],
}, index=['Climate', 'Housing', 'Health', 'Crime', 'Transportation',
          'Education', 'Arts', 'Recreation', 'Economy'])

print(loadings, '\n')
for pc in loadings.columns:
    top = loadings[pc].abs().idxmax()
    print(f'{pc}: strongest variable = {top} (loading = {loadings.loc[top, pc]})')

loadings.abs().plot(kind='bar', figsize=(10, 5))
plt.title('Absolute loadings of each variable on PC1, PC2, PC3')
plt.ylabel('|loading|'); plt.tight_layout(); plt.show()

print('\nConclusion: PC1 is dominated by "Arts" (0.985), followed by Health (0.782) and Transportation (0.585).')
```
