```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'Brightness': [40, 50, 60, 10, 70, 60, 25],
    'Saturation': [20, 50, 90, 25, 70, 10, 80],
    'Class':      ['Red', 'Blue', 'Blue', 'Red', 'Blue', 'Red', 'Blue']
})

new_entry = np.array([20, 35])   # new point: Brightness=20, Saturation=35 (change as needed)
K = 3

# Euclidean distance: sqrt((x1-x2)^2 + (y1-y2)^2)
df['Distance'] = np.sqrt((df['Brightness'] - new_entry[0])**2 +
                         (df['Saturation'] - new_entry[1])**2)
df_sorted = df.sort_values('Distance').reset_index(drop=True)
print(df_sorted)

nearest = df_sorted.head(K)
print(f'\n{K} nearest neighbours:\n', nearest)
print('\nPredicted class for', tuple(int(v) for v in new_entry), '->', nearest['Class'].value_counts().idxmax())

# Cross-check with scikit-learn
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=K).fit(df[['Brightness', 'Saturation']], df['Class'])
print('sklearn prediction:', knn.predict(pd.DataFrame([new_entry], columns=['Brightness', 'Saturation']))[0])
```
