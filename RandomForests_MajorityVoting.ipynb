```python
from collections import Counter


tree_votes = ['Apple', 'Apple', 'Banana']
counts = Counter(tree_votes)
print('Votes:', dict(counts))
print('Final class (majority voting):', counts.most_common(1)[0][0])


import numpy as np
from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

iris = load_iris()
rf = RandomForestClassifier(n_estimators=5, random_state=0).fit(iris.data, iris.target)
sample = iris.data[[0]]
votes = [str(iris.target_names[int(t.predict(sample)[0])]) for t in rf.estimators_]
print('\nIndividual tree votes:', votes)
print('Forest prediction    :', iris.target_names[rf.predict(sample)[0]])
```
